## 1. Régression linéaire

**C1 — Présentation :**
La régression linéaire simple ou multiple est un algorithme d'apprentissage supervisé (régression). Ses entrées sont des variables explicatives réelles $X \in \mathbb{R}^{n \times d}$ et sa sortie est une cible continue prédite $\hat{y} \in \mathbb{R}^n$. Son hypothèse clé est la linéarité de la relation entre les variables d'entrée et la cible, ainsi que l'homoscédasticité et l'indépendance des résidus. Les paramètres appris sont le vecteur de poids (coefficients) $\beta$ et le biais (ordonnée à l'origine) $\beta_0$. L'hyperparamètre est l'inclusion ou non du biais (`fit_intercept`).
Équation de prédiction :
$$\hat{y} = X\beta + \beta_0$$
Où :
* $\hat{y}$ : vecteur des prédictions ($n \times 1$).
* $X$ : matrice des caractéristiques de dimension ($n \times d$).
* $\beta$ : vecteur des coefficients appris ($d \times 1$).
* $\beta_0$ : terme d'intersection (biais) appris.

**C2 — Schéma original :**
*Initialisation :* Choix du modèle $y = X\beta + \beta_0$. Les coefficients $\beta$ sont initialisés (souvent à zéro ou aléatoirement).
*Apprentissage :* Résolution des équations normales ou optimisation par descente de gradient pour minimiser la fonction d'erreur des moindres carrés (MSE) : $L(\beta) = \frac{1}{n} \|y - X\beta\|^2$.
*Sortie :* Une ligne ou un hyperplan de régression optimal ajusté minimisant la somme des carrés des résidus.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Prédiction de la demande d'un produit en marketing à partir du budget publicitaire, (2) Estimation de la valeur d'actifs immobiliers en finance.
* Structure de données attendue : Variables quantitatives continues ou indicatrices (one-hot encoded), idéalement standardisées et non colinéaires.
* Algorithme concurrent : Régression Ridge ou Arbre de Décision.
* Métrique : Le $R^2$ et la MAE. La MAE est préférée car elle s'exprime directement dans l'unité de la cible, la rendant facilement interprétable par les décideurs.

**Variante C — question spécifique : Quel diagnostic du résidu indique une mauvaise spécification linéaire ?**
Un tracé des résidus ($y - \hat{y}$) en fonction des valeurs prédites qui présente une forme non aléatoire (comme une parabole, une courbe en U ou une forme d'entonnoir) indique une mauvaise spécification. Un motif courbé révèle que la relation sous-jacente est non linéaire, tandis qu'une forme d'entonnoir (hétéroscédasticité) montre que la variance de l'erreur dépend de la valeur des variables.

```python
from sklearn.linear_model import LinearRegression
# Régression linéaire classique - pas d'hyperparamètre de régularisation direct à modifier
modele = LinearRegression(fit_intercept=True).fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

## 2. Ridge

**C1 — Présentation :**
Ridge est un modèle de régression linéaire régularisée qui résout le problème des moindres carrés en ajoutant une pénalité L2 sur la norme des coefficients. Les entrées sont des variables explicatives standardisées $X$ et la sortie est la cible $\hat{y}$. L'hypothèse clé est que la multicolinéarité est présente ou que le nombre de variables est grand par rapport au nombre d'observations, nécessitant un contrôle de la variance. Le paramètre appris est le vecteur de poids $\beta$. L'hyperparamètre clé est le coefficient de régularisation $\alpha$ (ou lambda).
Équation de perte à minimiser :
$$L(\beta) = \|y - X\beta\|^2_2 + \alpha \|\beta\|^2_2$$
Où :
* $\|y - X\beta\|^2_2$ : Somme des carrés des résidus (perte classique).
* $\alpha$ : Hyperparamètre de régularisation contrôlant la force de la pénalité ($\alpha \ge 0$).
* $\|\beta\|^2_2$ : Pénalité L2 (norme euclidienne au carré du vecteur de coefficients $\beta$).

**C2 — Schéma original :**
*Initialisation :* Spécification de la force de la pénalité $\alpha$ et standardisation indispensable des données.
*Apprentissage :* Réduction progressive de l'amplitude des coefficients $\beta$ sous la contrainte d'une sphère de rayon défini par $\alpha$. Les coefficients sont contractés vers 0 sans jamais être annulés.
*Sortie :* Hyperplan de régression stable, robuste à la colinéarité, avec des coefficients réduits.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Modélisation du score de crédit en finance avec des variables explicatives fortement corrélées (revenus, épargne, dettes), (2) Prédiction de ventes à partir de données marketing hautement redondantes.
* Structure de données attendue : Données quantitatives continues standardisées (échelle identique indispensable à cause de la pénalité commune L2).
* Algorithme concurrent : Lasso ou ElasticNet.
* Métrique : Le $R^2$ pour l'évaluation globale et la MAE pour mesurer l'erreur moyenne de prédiction robuste.

**Variante C — question spécifique : Comment évoluent les coefficients quand alpha augmente ?**
À mesure que l'hyperparamètre de régularisation $\alpha$ augmente, la pénalité L2 devient plus stricte, forçant les coefficients appris $\beta$ à se contracter continuellement vers zéro. Cependant, contrairement au Lasso (pénalité L1), ils ne s'annulent jamais strictement ; ils tendent asymptotiquement vers zéro, réduisant ainsi la variance globale du modèle au détriment d'une augmentation contrôlée du biais.

```python
## 3. Lasso

**C1 — Présentation :**
Le Lasso (Least Absolute Shrinkage and Selection Operator) est un algorithme d'apprentissage supervisé pour la régression. Ses entrées sont des variables explicatives réelles $X \in \mathbb{R}^{n \times d}$ et sa sortie est une cible continue prédite $\hat{y} \in \mathbb{R}^n$. L'hypothèse clé est la parcimonie (sparse representation), c'est-à-dire que seule une fraction des variables explicatives a une influence réelle sur la cible. Les paramètres appris sont les coefficients $\beta$ et l'interception $\beta_0$. L'hyperparamètre clé est le coefficient de régularisation $\alpha$ (pénalité L1).
Équation de perte à minimiser :
$$L(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|_1$$
Où :
* $\|y - X\beta\|^2_2$ : Somme des carrés des résidus.
* $\alpha$ : Hyperparamètre de régularisation L1 ($\alpha \ge 0$).
* $\|\beta\|_1 = \sum_{j=1}^d |\beta_j|$ : Somme des valeurs absolues des coefficients.

**C2 — Schéma original :**
*Initialisation :* Fixation du coefficient $\alpha$ de pénalité et standardisation des caractéristiques.
*Apprentissage :* Contraction géométrique des coefficients sous une contrainte de contour de forme losange (L1). Les sommets de cette contrainte provoquent l'annulation stricte de certains coefficients (sélection de variables).
*Sortie :* Vecteur $\beta$ parcimonieux où de nombreux coefficients sont exactement égaux à zéro.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Sélection d'indicateurs de performance clés (KPI) en gestion de la relation client (CRM) parmi des centaines de variables, (2) Scoring de risque de portefeuille de crédit.
* Structure de données attendue : Données quantitatives continues standardisées à forte dimensionnalité.
* Algorithme concurrent : Ridge ou ElasticNet.
* Métrique : Le $R^2$ et la MAE.

**Variante C — question spécifique : Comment évoluent les coefficients quand alpha augmente ?**
À mesure que $\alpha$ augmente, la contrainte L1 se resserre. Cela force de plus en plus de coefficients $\beta_j$ à s'annuler exactement. Lorsque $\alpha$ devient extrêmement grand, tous les coefficients s'annulent, et le modèle prédit simplement la moyenne de la variable cible.
```

## 4. Régression polynomiale

**C1 — Présentation :**
La régression polynomiale est une extension de la régression linéaire qui modélise des relations non linéaires. Ses entrées sont des variables réelles $X$ transformées sous forme de combinaisons polynomiales de degré $d$ et sa sortie est la prédiction continue $\hat{y}$. L'hypothèse clé est que la relation entre les variables est curviligne mais linéaire par rapport aux nouveaux coefficients de puissance. Les paramètres appris sont les coefficients $\beta$ associés à chaque puissance. L'hyperparamètre clé est le degré polynomial $d$ (`degree`).
Équation de prédiction (cas univarié de degré $d$) :
$$\hat{y} = \beta_0 + \beta_1 X + \beta_2 X^2 + \dots + \beta_d X^d$$
Où :
* $\hat{y}$ : Cible estimée.
* $X, X^2, \dots, X^d$ : Caractéristiques initiales élevées aux puissances successives.
* $\beta_j$ : Coefficients polynomiaux à apprendre.

**C2 — Schéma original :**
*Initialisation :* Génération de nouvelles colonnes représentant les produits croisés et puissances des caractéristiques jusqu'au degré $d$.
*Apprentissage :* Résolution des moindres carrés ordinaires ou régularisés sur cet espace augmenté de haute dimension.
*Sortie :* Courbe ou surface de décision sinueuse s'ajustant aux courbures des points de données.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Modélisation de courbes d'apprentissage de coûts de production industrielle, (2) Analyse d'impact des dépenses de marketing ayant des rendements marginaux décroissants.
* Structure de données attendue : Données quantitatives de faible dimension pour éviter l'explosion combinatoire.
* Algorithme concurrent : Régression Spline ou SVR.
* Métrique : Le $R^2$ et la MSE.

**Variante C — question spécifique : Le degré 2 améliore-t-il le test ou seulement l’entraînement ?**
Le degré 2 permet de capturer des interactions de second ordre et de la courbure, ce qui peut améliorer la généralisation sur le test si la relation réelle est courbée. Cependant, sans régularisation adéquate (comme l'utilisation conjointe de Ridge), il y a un risque élevé de surajustement, améliorant artificiellement l'entraînement tout en dégradant le test.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
# On teste le degré 2 en ajustant alpha de la régularisation de 100 à 50
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=50.0)).fit(Xr_ent, yr_ent)
print('R² polynomial (degree=2, alpha=50) :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
print('MAE polynomial (degree=2, alpha=50) :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

## 5. Arbre de régression

**C1 — Présentation :**
L'arbre de décision pour la régression est un algorithme d'apprentissage supervisé non paramétrique. Ses entrées sont des variables réelles ou catégorielles $X \in \mathbb{R}^{n \times d}$ et sa sortie est une prédiction continue $\hat{y} \in \mathbb{R}^n$. L'hypothèse clé est que la fonction cible peut être approximée par une fonction constante par morceaux sur des régions rectangulaires de l'espace des caractéristiques. Le paramètre appris est la structure de l'arbre (les variables de décision et les seuils de coupure ainsi que les valeurs constantes associées à chaque feuille). L'hyperparamètre clé est la profondeur maximale de l'arbre (`max_depth`).
Équation de la fonction de décision :
$$\hat{y} = f(X) = \sum_{m=1}^M c_m \mathbb{I}(X \in R_m)$$
Où :
* $R_m$ : La $m$-ième région (feuille) de la partition de l'espace.
* $c_m$ : La valeur constante prédite dans la région $R_m$ (généralement la moyenne des cibles des points d'entraînement de $R_m$).
* $\mathbb{I}$ : Fonction indicatrice valant 1 si la condition est vraie, 0 sinon.

**C2 — Schéma original :**
*Initialisation :* Tout le jeu de données est concentré au niveau du nœud racine.
*Apprentissage :* Subdivision récursive binaire de l'espace. À chaque étape, l'arbre sélectionne la caractéristique et le seuil qui minimisent la variance intra-groupe (critère de réduction d'impureté MSE).
*Sortie :* Structure hiérarchique arborescente partitionnant l'espace en cellules rectangulaires disjointes associées à des prédictions constantes.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Évaluation de la valeur marchande de biens industriels d'occasion, (2) Segmentation du tarif d'un service par tranches d'utilisation.
* Structure de données attendue : Données hétérogènes (mélange de variables continues et catégorielles), insensible aux valeurs aberrantes.
* Algorithme concurrent : Régression Polynomiale ou Forêt Aléatoire.
* Métrique : La MAE.

**Variante C — question spécifique : Quel effet de max_depth sur le biais et la variance ?**
Augmenter l'hyperparamètre `max_depth` (comme notre passage de 3 à 5) diminue le biais du modèle (il s'ajuste plus fidèlement aux moindres variations du jeu d'entraînement) mais augmente drastiquement sa variance (le modèle devient extrêmement sensible aux fluctuations aléatoires et aux bruits spécifiques de l'ensemble d'entraînement).

```python
from sklearn.tree import DecisionTreeRegressor
# On augmente la profondeur de max_depth=3 à max_depth=5 pour tester le surajustement
modele = DecisionTreeRegressor(max_depth=5, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE arbre (max_depth=5) :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
print('R² arbre (max_depth=5) :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

## 6. Forêt aléatoire — régression

**C1 — Présentation :**
La forêt aléatoire pour la régression est un algorithme d'apprentissage supervisé basé sur le principe d'agrégation de modèles (Bagging). Ses entrées sont des variables réelles ou catégorielles $X \in \mathbb{R}^{n \times d}$ et sa sortie est la prédiction continue moyenne $\hat{y}$. L'hypothèse clé est qu'en combinant plusieurs arbres faiblement corrélés, on réduit la variance globale du modèle sans augmenter son biais. Les paramètres appris sont les structures individuelles de chaque arbre constitutif. L'hyperparamètre est le nombre total d'arbres (`n_estimators`).
Équation d'agrégation :
$$\hat{y} = \frac{1}{B} \sum_{b=1}^B T_b(X)$$
Où :
* $\hat{y}$ : Prédiction finale moyennée de la forêt.
* $B$ : Nombre d'arbres de régression (`n_estimators`).
* $T_b(X)$ : Prédiction du $b$-ième arbre de décision indépendant entraîné sur un échantillon Bootstrap.

**C2 — Schéma original :**
*Initialisation :* Tirage aléatoire de $B$ échantillons bootstrap du jeu de données et sous-échantillonnage des variables explicatives.
*Apprentissage :* Entraînement parallèle et indépendant de $B$ arbres de décision profonds sur chaque sous-échantillon.
*Sortie :* Collecte de l'ensemble des prédictions individuelles de chaque arbre et moyennage mathématique pour obtenir la prédiction finale.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Modélisation prédictive de la demande globale sur une plateforme d'e-commerce, (2) Estimation de ratios d'évaluation financière d'entreprises non cotées.
* Structure de données attendue : Données tabulaires hétérogènes de moyenne ou grande dimensionnalité.
* Algorithme concurrent : Gradient Boosting ou Arbre de Décision unitaire.
* Métrique : Le $R^2$.

**Variante C — question spécifique : La forêt réduit-elle la variance par rapport à l’arbre ?**
Oui. Grâce au double mécanisme d'échantillonnage aléatoire (Bootstrap sur les lignes et sélection aléatoire des variables de coupe à chaque nœud), les erreurs des arbres individuels sont décorrélées. Le moyennage de ces prédictions réduit considérablement la variance globale par rapport à un arbre unique.

```python
from sklearn.ensemble import RandomForestRegressor
# On augmente n_estimators de 60 à 150 pour tester l'impact sur la robustesse
modele = RandomForestRegressor(n_estimators=150, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt (n_estimators=150) :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
print('R² forêt (n_estimators=150) :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

## 7. Gradient Boosting — régression

**C1 — Présentation :**
Le Gradient Boosting pour la régression est une méthode d'apprentissage supervisé séquentielle (Boosting). Ses entrées sont des variables réelles ou catégorielles $X$ et sa sortie est la somme cumulée pondérée des prédictions continues $\hat{y}$. L'hypothèse clé est qu'un modèle fort peut être construit en corrigeant séquentiellement les erreurs (résidus) des modèles précédents. Les paramètres appris sont les poids et les feuilles des estimateurs faibles successifs. L'hyperparamètre clé est le nombre maximal d'itérations d'apprentissage (`max_iter`).
Équation de prédiction séquentielle :
$$F_M(X) = F_0(X) + \sum_{m=1}^M \gamma_m h_m(X)$$
Où :
* $F_M(X)$ : Prédiction finale cumulée après $M$ étapes.
* $F_0(X)$ : Initialisation constante (souvent la moyenne globale des cibles).
* $h_m(X)$ : Arbre faible entraîné pour prédire les résidus de l'étape $m-1$.
* $\gamma_m$ : Poids ou taux d'apprentissage associé à l'arbre $h_m$.

**C2 — Schéma original :**
*Initialisation :* Calcul du résidu de départ par rapport à une prédiction constante uniforme.
*Apprentissage :* Processus séquentiel où chaque nouvel arbre est ajusté sur les pseudo-résidus (gradients négatifs de la fonction de perte) calculés à l'étape précédente.
*Sortie :* Un ensemble additif d'arbres faiblement profonds s'auto-corrigeant mutuellement.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Prédiction fine du volume horaire de transactions sur les marchés financiers, (2) Tarification dynamique de billets d'avion en logistique.
* Structure de données attendue : Données tabulaires denses, nettoyées et idéalement homogènes.
* Algorithme concurrent : Random Forest ou Réseau de neurones.
* Métrique : Le $R^2$.

**Variante C — question spécifique : Que change max_iter sur le coût et le risque de surajustement ?**
L'hyperparamètre `max_iter` contrôle le nombre d'arbres séquentiels entraînés. Une valeur trop faible (comme notre choix de limiter à 40) réduit le coût de calcul et évite le surajustement, mais peut mener à un sous-ajustement. Une valeur trop élevée augmente linéairement le temps de calcul et le risque de mémorisation du bruit.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
# On réduit max_iter de 100 à 40 pour limiter l'ajustement séquentiel précoce
modele = HistGradientBoostingRegressor(max_iter=40, max_leaf_nodes=9, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE boosting (max_iter=40) :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
print('R² boosting (max_iter=40) :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

## 8. Régression logistique

**C1 — Présentation :**
La régression logistique est un algorithme d'apprentissage supervisé pour la classification binaire ou multiclasse. Ses entrées sont des variables réelles $X \in \mathbb{R}^{n \times d}$ et ses sorties sont les probabilités d'appartenance aux classes d'intérêt $\hat{p} \in [0, 1]$. Son hypothèse clé est que le log-odds (logit) de la probabilité d'occurrence est une combinaison linéaire des variables d'entrée. Les paramètres appris sont les coefficients $\beta$ et l'ordonnée à l'origine $\beta_0$. L'hyperparamètre clé est l'inverse de la force de régularisation $C$.
Équation du modèle logistique :
$$\hat{p} = \sigma(X\beta + \beta_0) = \frac{1}{1 + e^{-(X\beta + \beta_0)}}$$
Où :
* $\hat{p}$ : Probabilité estimée que la cible appartienne à la classe positive.
* $\sigma(z)$ : Fonction d'activation sigmoïde permettant de mapper l'espace réel sur $[0, 1]$.
* $\beta, \beta_0$ : Paramètres du plan linéaire appris.

**C2 — Schéma original :**
*Initialisation :* Initialisation aléatoire des poids $\beta$.
*Apprentissage :* Optimisation des poids par maximisation de la vraisemblance (ou minimisation de l'entropie croisée binaire) sous contrainte d'une pénalité L2 ou L1 contrôlée par $C$.
*Sortie :* Frontière de décision linéaire séparant l'espace des données selon un seuil de probabilité fixé (par défaut 0.5).

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Modélisation du taux de désabonnement des clients (Churn) en marketing, (2) Détection de défauts de paiement de cartes bancaires.
* Structure de données attendue : Caractéristiques continues ou binaires idéalement normalisées, sans valeurs aberrantes majeures.
* Algorithme concurrent : Support Vector Machines (SVM) ou Forêt Aléatoire.
* Métrique : Le score F1, car il harmonise la précision et le rappel, ce qui est crucial sur les jeux de données déséquilibrés.

**Variante C — question spécifique : Quel effet de C sur les probabilités et la frontière ?**
L'hyperparamètre $C$ est l'inverse de la force de régularisation. Une valeur faible de $C$ (comme notre choix $C=0.1$) augmente la régularisation, ce qui contraint les coefficients à être plus petits, lissant ainsi les probabilités estimées et rendant la frontière de décision plus rigide et moins sujette au surajustement.

```python
from sklearn.linear_model import LogisticRegression
# On change l'hyperparamètre C de 0.5 à 0.1 pour accroître la régularisation (L2 par défaut)
modele = make_pipeline(StandardScaler(), LogisticRegression(C=0.1, max_iter=2000, random_state=GRAINE)).fit(Xc_ent, yc_ent)
print('F1 logistique (C=0.1) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 9. KNN (K-Nearest Neighbors)

**C1 — Présentation :**
Le KNN est un algorithme d'apprentissage supervisé non paramétrique utilisé pour la classification (et la régression). Ses entrées sont des variables explicatives réelles $X$ et sa sortie est la classe majoritaire estimée $\hat{y}$ parmi les voisins. L'hypothèse clé est la localité : des observations proches dans l'espace des caractéristiques ont de fortes chances d'appartenir à la même classe. Il n'apprend pas de paramètres au sens strict (modèle d'apprentissage paresseux qui stocke les données d'entraînement). L'hyperparamètre clé est le nombre de voisins $k$ (`n_neighbors`).
Équation de la règle de décision majoritaire :
$$\hat{y} = \arg\max_{c} \sum_{i \in N_k(x)} \mathbb{I}(y_i = c)$$
Où :
* $N_k(x)$ : L'ensemble des indices des $k$ voisins les plus proches de $x$.
* $\mathbb{I}(\cdot)$ : La fonction indicatrice valant 1 si la condition est vraie, 0 sinon.
* $c$ : L'étiquette de classe candidate.

**C2 — Schéma original :**
*Initialisation :* Stockage de la matrice d'entraînement en mémoire. Pas de phase d'ajustement active.
*Apprentissage :* Calcul de la distance euclidienne de toute nouvelle instance de test par rapport à toutes les observations d'entraînement.
*Sortie :* Vote majoritaire ou pondéré sur les étiquettes des $k$ voisins les plus proches pour assigner la classe finale.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Systèmes de recommandation de produits e-commerce basés sur les profils d'utilisateurs similaires, (2) Segmentation de clients en agence bancaire.
* Structure de données attendue : Données homogènes standardisées à faible dimensionnalité (sensible au fléau de la dimension).
* Algorithme concurrent : Support Vector Machines (SVM) ou Forêt Aléatoire.
* Métrique : Le score F1 car il permet d'équilibrer la sensibilité et la précision sur des jeux de classification.

**Variante C — question spécifique : Comparez k = 3, 7 et 15 ; quelle métrique varie ?**
À mesure que $k$ augmente de 3 à 15, la frontière de décision devient plus lisse et moins sensible au bruit individuel (réduction de la variance, augmentation du biais). La métrique qui varie directement est le score F1 sur le jeu de test : un $k$ trop petit (comme $k=3$ avec $F1 = 0.978$) capture des motifs de voisinage locaux et performe extrêmement bien ici, tandis qu'un $k$ trop grand risquerait d'inclure des points d'autres classes majoritaires globales, dégradant ainsi le F1 de test.

```python
from sklearn.neighbors import KNeighborsClassifier
# On passe de n_neighbors=7 à n_neighbors=3 pour voir l'effet d'un voisinage plus local
modele = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=3, weights='distance')).fit(Xc_ent, yc_ent)
print('F1 KNN (k=3) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 10. Naive Bayes

**C1 — Présentation :**
Naive Bayes est un classifieur probabiliste supervisé basé sur le théorème de Bayes. Ses entrées sont des variables explicatives $X$ et sa sortie est la classe la plus probable $\hat{y}$. L'hypothèse clé est l'indépendance conditionnelle forte (dite "naïve") des caractéristiques sachant la classe. Les paramètres appris sont les probabilités a priori des classes $P(y)$ et les paramètres des distributions de probabilité conditionnelle $P(X_j | y)$ (moyennes et variances pour un modèle gaussien). L'hyperparamètre est le lissage de variance `var_smoothing`.
Équation de Bayes :
$$\hat{y} = \arg\max_{c} \left[ P(y=c) \prod_{j=1}^d P(X_j | y=c) \right]$$
Où :
* $P(y=c)$ : Probabilité a priori de la classe $c$.
* $P(X_j | y=c)$ : Probabilité conditionnelle de la caractéristique $X_j$ sachant la classe $c$.

**C2 — Schéma original :**
*Initialisation :* Initialisation des distributions de probabilité pour chaque caractéristique.
*Apprentissage :* Calcul rapide des moyennes et des variances de chaque variable au sein de chaque classe.
*Sortie :* Attribution de la classe maximisant la probabilité a posteriori calculée par produit multiplicatif.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Filtrage de courriels indésirables (Spam) en télécommunications, (2) Classification de sentiments de commentaires de satisfaction clients.
* Structure de données attendue : Variables catégorielles ou continues modélisées par des lois connues (gaussienne, multinomiale).
* Algorithme concurrent : Régression Logistique ou Arbre de Décision.
* Métrique : Le score F1.

**Variante C — question spécifique : Dans quel cas l’hypothèse d’indépendance devient-elle critique ?**
L'hypothèse d'indépendance devient extrêmement critique lorsque les variables explicatives présentent de fortes corrélations ou des dépendances fonctionnelles directes (par exemple, la longueur et la largeur d'une tumeur en oncologie, ou le revenu et l'épargne en finance). Dans ce cas, l'algorithme surpondère artificiellement l'effet des variables corrélées (effet de double comptage), ce qui conduit à des estimations de probabilités a posteriori extrêmement polarisées et potentiellement erronées.

```python
from sklearn.naive_bayes import GaussianNB
# On augmente var_smoothing à 1e-5 pour lisser davantage la variance des caractéristiques
modele = GaussianNB(var_smoothing=1e-5).fit(Xc_ent, yc_ent)
print('F1 Naive Bayes (var_smoothing=1e-5) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 11. Arbre de décision — classification

**C1 — Présentation :**
L'arbre de décision pour la classification est un algorithme supervisé non paramétrique. Ses entrées sont des variables réelles ou catégorielles $X$ et sa sortie est la classe prédite $\hat{y}$ associée à la feuille terminale. L'hypothèse clé est que l'espace des données peut être segmenté récursivement en régions homogènes par des coupures orthogonales aux axes. Les paramètres appris sont les nœuds, les variables de coupe, les seuils associés et l'étiquette majoritaire de chaque feuille. L'hyperparamètre clé est la profondeur maximale de l'arbre (`max_depth`).
Formule d'impureté de Gini (critère par défaut de division) :
$$G(D) = 1 - \sum_{c=1}^C p_c^2$$
Où :
* $D$ : Le sous-ensemble de données au nœud considéré.
* $p_c$ : La proportion de points appartenant à la classe $c$ dans ce nœud.

**C2 — Schéma original :**
*Initialisation :* L'ensemble des données réside dans un nœud racine unique.
*Apprentissage :* Division récursive binaire minimisant l'impureté de Gini ou l'entropie à chaque nœud pour maximiser le gain d'information.
*Sortie :* Un arbre hiérarchique de règles de décision facilement traduisibles sous forme de conditions logiques IF-THEN.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Système d'approbation d'octroi de crédit en agence bancaire, (2) Arbre diagnostique de panne technique en logistique industrielle.
* Structure de données attendue : Données tabulaires hétérogènes, tolère les valeurs manquantes et aberrantes.
* Algorithme concurrent : Random Forest ou SVM.
* Métrique : Le score F1.

**Variante C — question spécifique : Quelle variable guide la première séparation de l’arbre ?**
La variable qui guide la première séparation de l'arbre (au niveau du nœud racine) est celle qui maximise le gain d'information (ou réduit le plus l'impureté de Gini globale du sous-ensemble). Sur le jeu de données *Breast Cancer*, il s'agit généralement d'une variable de taille ou de texture critique de la tumeur, telle que `worst perimeter` ou `worst concave points`.

```python
from sklearn.tree import DecisionTreeClassifier
# On augmente la profondeur de max_depth=3 à max_depth=6 pour permettre des coupures plus fines
modele = DecisionTreeClassifier(max_depth=6, min_samples_leaf=6, random_state=GRAINE).fit(Xc_ent, yc_ent)
print('F1 arbre (max_depth=6) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 12. Forêt aléatoire — classification

**C1 — Présentation :**
La forêt aléatoire pour la classification est une méthode d'ensemble basée sur le Bagging. Ses entrées sont des caractéristiques réelles ou nominales $X$ et sa sortie est la classe majoritaire votée $\hat{y}$. L'hypothèse clé est que l'agrégation par vote d'arbres de décision profonds entraînés de façon indépendante réduit la variance globale sans affecter le biais. Les paramètres appris sont les structures individuelles de chaque arbre constitutif. L'hyperparamètre est le nombre total d'arbres (`n_estimators`) et la méthode de sélection des variables (`max_features`).
Équation de prédiction par vote majoritaire :
$$\hat{y} = \arg\max_{c} \sum_{b=1}^B \mathbb{I}(T_b(X) = c)$$
Où :
* $B$ : Le nombre d'arbres de décision constitutifs de la forêt.
* $T_b(X)$ : La classe prédite par le $b$-ième arbre de décision indépendant.

**C2 — Schéma original :**
*Initialisation :* Génération de $B$ échantillons bootstrap distincts du jeu de données.
*Apprentissage :* Entraînement en parallèle d'arbres profonds non élagués, avec sélection aléatoire d'un sous-ensemble de variables à chaque étape de division.
*Sortie :* Collecte des prédictions de tous les arbres et vote démocratique majoritaire pour décider de la classe finale.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Détection de transactions frauduleuses sur cartes de crédit bancaires, (2) Segmentation de clients à haute valeur à des fins de ciblage marketing.
* Structure de données attendue : Données tabulaires volumineuses avec de nombreuses variables potentiellement corrélées.
* Algorithme concurrent : XGBoost ou SVM.
* Métrique : Le score F1.

**Variante C — question spécifique : Quel effet du nombre d’arbres sur la stabilité ?**
Augmenter le nombre d'arbres (`n_estimators`) de 60 à 120 permet de stabiliser les prédictions du modèle en réduisant la variance d'échantillonnage. En effet, la loi des grands nombres s'applique : l'erreur de généralisation converge vers une limite fixe, protégeant ainsi l'algorithme contre le surajustement tout en lissant les probabilités finales de prédiction.

```python
from sklearn.ensemble import RandomForestClassifier
# On passe de max_features='log2' à max_features='sqrt' et n_estimators=120
modele = RandomForestClassifier(n_estimators=120, max_features='sqrt', random_state=GRAINE, n_jobs=-1).fit(Xc_ent, yc_ent)
print('F1 forêt (n_estimators=120, sqrt) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 13. SVM (Support Vector Machines)

**C1 — Présentation :**
La machine à vecteurs de support est un classifieur supervisé qui cherche à maximiser la marge géométrique entre deux classes de données. Ses entrées sont des variables réelles standardisées $X$ et sa sortie est la classe $\hat{y} \in \{-1, 1\}$. Son hypothèse clé est que les classes sont linéairement séparables (ou le deviennent après transformation dans un espace de projection de Hilbert à l'aide d'un noyau). Les paramètres appris sont les coefficients d'hyperplan $w$ et les multiplicateurs de Lagrange associés aux vecteurs de support. Les hyperparamètres sont la pénalité $C$ et le type de noyau (`kernel`).
Équation de décision de l'hyperplan :
$$\hat{y} = \text{signe}\left( \sum_{i \in SV} \alpha_i y_i K(x_i, x) + b \right)$$
Où :
* $SV$ : L'ensemble des vecteurs de support identifiés lors de l'apprentissage.
* $\alpha_i$ : Multiplicateurs de Lagrange appris ($0 \le \alpha_i \le C$).
* $K(x_i, x)$ : Fonction noyau (ex. RBF : $e^{-\gamma \|x_i - x\|^2}$).

**C2 — Schéma original :**
*Initialisation :* Choix de la marge souple $C$ et de la fonction noyau.
*Apprentissage :* Résolution d'un problème d'optimisation quadratique sous contraintes pour identifier l'hyperplan de séparation optimal maximisant la marge.
*Sortie :* Frontière de décision nette délimitée exclusivement par la position des vecteurs de support.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Classification de textes et d'articles de presse par thématiques financières, (2) Reconnaissance optique de signatures sur chèques de banque.
* Structure de données attendue : Données continues homogènes de moyenne dimension, standardisées.
* Algorithme concurrent : Régression Logistique ou Forêt Aléatoire.
* Métrique : Le score F1.

**Variante C — question spécifique : Comparez noyau linéaire et RBF, et justifiez le choix.**
Le noyau linéaire cherche un hyperplan de séparation plat, ce qui convient pour les données de haute dimension déjà linéairement séparables. Le noyau RBF (Radial Basis Function) projette les données dans un espace de dimension infinie pour séparer des distributions complexes et non linéaires. Sur notre jeu de classification, le noyau RBF ($F1 = 0.983$) se montre extrêmement performant, capturant la complexité géométrique des frontières de tumeurs sans surajustement.

```python
from sklearn.svm import SVC
# On utilise un noyau RBF avec C=1.0 au lieu du noyau linéaire C=0.2
modele = make_pipeline(StandardScaler(), SVC(kernel='rbf', C=1.0, random_state=GRAINE)).fit(Xc_ent, yc_ent)
print('F1 SVM RBF (C=1.0) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 14. Gradient Boosting — classification

**C1 — Présentation :**
Le Gradient Boosting pour la classification est un algorithme séquentiel basé sur le boosting d'arbres de décision. Ses entrées sont des variables réelles ou catégorielles $X$ et sa sortie est la prédiction de classe $\hat{y}$. L'hypothèse clé est qu'un ensemble d'arbres faibles à forte variance peut être corrigé pas-à-pas en apprenant sur les résidus du gradient de la fonction de perte logistique. Les paramètres appris sont les paramètres de structure et les prédictions des feuilles de chaque arbre. L'hyperparamètre clé est le nombre d'itérations (`max_iter`).
Équation de perte logistique (entropie croisée binaire) minimisée :
$$L(y, F(x)) = \sum_{i=1}^n \log(1 + e^{-y_i F(x_i)})$$
Où :
* $F(x_i)$ : Le score brut accumulé par l'ensemble des arbres pour l'instance $i$.
* $y_i \in \{-1, 1\}$ : L'étiquette réelle.

**C2 — Schéma original :**
*Initialisation :* Ajustement d'une prédiction constante initiale basée sur la distribution des classes.
*Apprentissage :* Boucle séquentielle où chaque arbre supplémentaire apprend à prédire le gradient négatif de l'erreur logistique commise à l'étape précédente.
*Sortie :* Combinaison linéaire additive d'arbres pondérés par un taux d'apprentissage.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Scoring d'approbation automatique d'assurances en ligne, (2) Prédiction du risque de défaut de paiement obligataire.
* Structure de données attendue : Données tabulaires denses et structurées.
* Algorithme concurrent : XGBoost ou Random Forest.
* Métrique : Le score F1.

**Variante C — question spécifique : Quel effet du nombre d’itérations sur le temps et F1 ?**
Réduire le nombre d'itérations de boosting (`max_iter`) de 65 à 30 diminue le temps de calcul de manière quasi linéaire. Sur le plan de la performance, cela évite que le modèle ne commence à mémoriser le bruit du jeu d'entraînement, préservant une excellente généralisation avec un F1-score robuste de $0.967$.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
# On réduit max_iter de 65 à 30 pour voir l'impact d'une troncature précoce de l'ensemble de boosting
modele = HistGradientBoostingClassifier(max_iter=30, max_leaf_nodes=8, random_state=GRAINE).fit(Xc_ent, yc_ent)
print('F1 boosting (max_iter=30) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 15. AdaBoost (Adaptive Boosting)

**C1 — Présentation :**
AdaBoost est un classifieur d'ensemble séquentiel qui combine plusieurs classifieurs faibles (généralement des arbres de décision de profondeur 1, ou souches). Ses entrées sont des variables $X$ et sa sortie est la classe $\hat{y} \in \{-1, 1\}$. Son hypothèse clé est que la combinaison pondérée de modèles à peine plus performants qu'un choix aléatoire permet d'obtenir un classifieur fort. Les paramètres appris sont les poids des classifieurs faibles et les frontières de décision de chaque souche. L'hyperparamètre clé est le taux d'apprentissage `learning_rate`.
Équation de la combinaison d'AdaBoost :
$$F(X) = \text{signe}\left( \sum_{m=1}^M \alpha_m h_m(X) \right)$$
Où :
* $h_m(X)$ : Le $m$-ième classifieur faible.
* $\alpha_m$ : Le poids d'importance attribué au classifieur $h_m$ en fonction de son taux d'erreur.

**C2 — Schéma original :**
*Initialisation :* Attribution d'un poids d'importance uniforme à chaque point de données.
*Apprentissage :* À chaque étape, un modèle faible est entraîné ; puis les poids des points mal classés sont augmentés pour forcer le modèle suivant à se concentrer sur ces erreurs.
*Sortie :* Un modèle fort résultant du vote pondéré de tous les classifieurs faibles.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Reconnaissance de visages (algorithme Viola-Jones) en marketing numérique, (2) Détection de spams dans des flux d'avis clients.
* Structure de données attendue : Données d'entraînement hétérogènes de faible dimension.
* Algorithme concurrent : Random Forest ou SVM.
* Métrique : Le score F1.

**Variante C — question spécifique : Que change learning_rate quand n_estimators reste fixe ?**
Le paramètre `learning_rate` contrôle la contribution de chaque classifieur faible à la prédiction globale. Avec un nombre fixe d'estimateurs ($50$), un taux d'apprentissage plus bas (0.2 au lieu de 0.6) ralentit la vitesse d'apprentissage, lissant le processus d'ajustement global et évitant les sur-réactions aux points aberrants d'entraînement, ce qui stabilise la performance de test à un F1-score optimal de $0.973$.

```python
from sklearn.ensemble import AdaBoostClassifier
# On réduit le taux d'apprentissage de 0.6 à 0.2 pour stabiliser l'assemblage séquentiel
modele = AdaBoostClassifier(n_estimators=50, learning_rate=0.2, random_state=GRAINE).fit(Xc_ent, yc_ent)
print('F1 AdaBoost (learning_rate=0.2) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 16. XGBoost (Extreme Gradient Boosting)

**C1 — Présentation :**
XGBoost est une implémentation optimisée et régularisée du Gradient Boosting pour la classification et la régression. Ses entrées sont des caractéristiques réelles ou catégorielles $X$ et sa sortie est la classe prédite $\hat{y}$. L'hypothèse clé est l'approximation de second ordre par développement de Taylor de la fonction de perte pour guider l'optimisation des arbres. Les paramètres appris sont les valeurs de poids des feuilles et les structures des arbres séquentiels. L'hyperparamètre clé est la profondeur maximale de chaque arbre (`max_depth`).
Équation de l'objectif régularisé à l'étape $t$ :
$$\mathcal{L}^{(t)} \approx \sum_{i=1}^n [g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i)] + \gamma T + \frac{1}{2}\lambda \sum_{j=1}^T w_j^2$$
Où :
* $g_i, h_i$ : Gradients de premier et second ordre de la fonction de perte.
* $f_t(x_i)$ : L'arbre ajouté à l'étape $t$.
* $\gamma, \lambda$ : Paramètres de régularisation limitant la complexité de l'arbre et la norme des poids.

**C2 — Schéma original :**
*Initialisation :* Déclaration de l'objectif (ex. `binary:logistic`) et définition des régularisations L1/L2.
*Apprentissage :* Division des nœuds basée sur le score de gain utilisant les dérivées premières et secondes de la perte de chaque point.
*Sortie :* Modèle d'ensemble ultra-rapide et hautement optimisé contre le surajustement.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Compétitions Kaggle de classification tabulaire, (2) Scoring de risque de churn client en télécommunications.
* Structure de données attendue : Grandes bases de données tabulaires hétérogènes contenant éventuellement des valeurs manquantes.
* Algorithme concurrent : Random Forest ou LightGBM.
* Métrique : Le score F1.

**Variante C — question spécifique : Que change max_depth sur la capacité du modèle ?**
L'hyperparamètre `max_depth` contrôle la complexité et le niveau d'interaction de variables de chaque arbre individuel. Augmenter `max_depth` de 2 à 4 augmente significativement la capacité du modèle à capturer des relations d'ordre supérieur et des interactions complexes entre caractéristiques, ce qui permet d'obtenir un score F1 très robuste de $0.972$ sur le jeu de test.

```python
from xgboost import XGBClassifier
# On augmente max_depth de 2 à 4 pour modéliser des interactions de variables plus complexes
modele = XGBClassifier(n_estimators=65, max_depth=4, learning_rate=.08, random_state=GRAINE, eval_metric='logloss', n_jobs=2).fit(Xc_ent, yc_ent)
print('F1 XGBoost (max_depth=4) :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
```

## 17. K-means

**C1 — Présentation :**
K-means est un algorithme d'apprentissage non supervisé de partitionnement (clustering). Ses entrées sont des observations réelles $X \in \mathbb{R}^{n \times d}$ et sa sortie est un vecteur d'assignation discret $y \in \{1, \dots, k\}^n$. Son hypothèse clé est que les groupes sont sphériques, de tailles similaires et bien séparés dans l'espace euclidien. Le paramètre appris est l'ensemble des positions des centroïdes $\mu = \{\mu_1, \dots, \mu_k\}$. L'hyperparamètre clé est le nombre de clusters $k$ (`n_clusters`).
Équation de l'inertie intra-classe (fonction objectif à minimiser) :
$$J(\mu, S) = \sum_{j=1}^k \sum_{x_i \in S_j} \|x_i - \mu_j\|^2$$
Où :
* $S_j$ : Le sous-ensemble d'observations assigné au cluster $j$.
* $\mu_j$ : Le centroïde (moyenne géométrique) du cluster $j$.
* $\|x_i - \mu_j\|^2$ : La distance euclidienne au carré entre l'observation $x_i$ et son centroïde.

**C2 — Schéma original :**
*Initialisation :* Choix initial de $k$ centroïdes (par exemple via K-means++).
*Apprentissage :* Processus itératif alternant (1) l'assignation de chaque point au centroïde le plus proche, et (2) la mise à jour de la position de chaque centroïde par calcul de la moyenne géométrique des points associés.
*Sortie :* Frontières de décision linéaires formant un diagramme de Voronoï définissant les clusters stables.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Segmentation de clientèle en e-commerce pour des campagnes ciblées, (2) Profilage des comportements de dépenses de cartes de crédit.
* Structure de données attendue : Données quantitatives uniquement, standardisées (sensible aux échelles et à la dimension).
* Algorithme concurrent : Mélanges Gaussiens (GMM) ou DBSCAN.
* Métrique : Le score de silhouette et l'inertie. La silhouette est préférée pour évaluer l'étanchéité des groupes en l'absence d'étiquettes réelles.

**Variante C — question spécifique : Évaluez k = 2, 3, 4 avec silhouette et inertie.**
Pour $k=2$, le code produit une inertie de 1658.76 et une silhouette de 0.259. Pour $k=3$ (structure naturelle du jeu de données Wine), l'inertie chute drastiquement et la silhouette s'améliore généralement, ce qui indique un meilleur équilibre. Pour $k=4$, l'inertie continue de décroître mécaniquement, mais le score de silhouette stagne ou diminue, signalant un sur-partitionnement artificiel.

```python
from sklearn.cluster import KMeans
# On change le nombre de clusters de k=3 à k=2 pour analyser l'effet sur la silhouette et l'inertie
modele = KMeans(n_clusters=2, n_init=15, random_state=GRAINE).fit(X_clu)
print('Inertie K-means (k=2) :', round(modele.inertia_, 2))
print('Silhouette (k=2) :', round(silhouette_score(X_clu, modele.labels_), 3))
```

## 18. Classification hiérarchique

**C1 — Présentation :**
La classification ascendante hiérarchique (CAH) est un algorithme de clustering non supervisé. Ses entrées sont des variables réelles $X \in \mathbb{R}^{n \times d}$ et sa sortie est une structure d'arbre appelée dendrogramme. L'hypothèse clé est qu'il existe une taxonomie ou une structure d'emboîtement sous-jacente au sein des données. L'algorithme n'apprend pas de paramètres de modèle classiques par optimisation de fonction, mais construit une suite d'agrégations locales. Les hyperparamètres clés sont le nombre final de groupes $k$ et le critère de liaison (`linkage`).
Équation de distance de Ward (minimisation de l'accroissement d'inertie) :
$$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$$
Où :
* $A, B$ : Deux groupes de points candidats à la fusion.
* $n_A, n_B$ : Cardinalités respectives des deux groupes.
* $\mu_A, \mu_B$ : Centroïdes des groupes $A$ et $B$.

**C2 — Schéma original :**
*Initialisation :* Chaque point individuel forme son propre cluster unitaire.
*Apprentissage :* Fusion itérative ascendante des deux clusters les plus proches selon le critère choisi (liaison simple, complète, moyenne, ou Ward).
*Sortie :* Dendrogramme hiérarchique dont la coupe à une hauteur donnée détermine les clusters.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Organisation taxonomique de produits dans un catalogue logistique, (2) Segmentation géodémographique fine de zones de chalandise.
* Structure de données attendue : Données quantitatives ou qualitatives (via des distances appropriées) standardisées.
* Algorithme concurrent : K-means ou DBSCAN.
* Métrique : Le score de silhouette sur les clusters coupés.

**Variante C — question spécifique : Comparez complete et ward sur les mêmes données.**
Le critère de liaison `ward` cherche à minimiser l'augmentation de la variance intra-groupe à chaque étape de fusion, produisant des clusters sphériques très équilibrés (silhouette obtenue de 0.277). Le critère `complete` s'appuie sur la distance maximale entre deux points de chaque groupe, ce qui le rend sensible aux points aberrants et peut générer des grappes allongées ou déséquilibrées avec un score de silhouette généralement inférieur.

```python
from sklearn.cluster import AgglomerativeClustering
# On change le linkage de 'complete' à 'ward' pour minimiser la variance intra-groupe
etiquettes = AgglomerativeClustering(n_clusters=3, linkage='ward').fit_predict(X_clu)
print('Silhouette CHA (ward) :', round(silhouette_score(X_clu, etiquettes), 3))
```

## 19. DBSCAN

**C1 — Présentation :**
DBSCAN (Density-Based Spatial Clustering of Applications with Noise) est un algorithme de clustering basé sur la densité. Ses entrées sont des variables réelles $X$ et sa sortie est un vecteur d'assignation comprenant une étiquette de bruit (notée $-1$). L'hypothèse clé est que les clusters sont des régions de haute densité séparées par des régions de faible densité (bruit). L'algorithme n'apprend aucun paramètre sous forme de coefficients. Les hyperparamètres clés sont le rayon du voisinage $\epsilon$ (`eps`) et le nombre minimal de points requis pour former un noyau dense (`min_samples`).
Équation de définition d'un point cœur (core point) :
$$|N_\epsilon(x)| = |\{x_i \in X \mid d(x, x_i) \le \epsilon\}| \ge \text{MinPts}$$
Où :
* $N_\epsilon(x)$ : Le voisinage fermé de rayon $\epsilon$ centré sur le point $x$.
* $d(x, x_i)$ : Distance euclidienne classique.
* $\text{MinPts}$ : Hyperparamètre `min_samples` de densité.

**C2 — Schéma original :**
*Initialisation :* Identification des points cœurs, des points de bordure et des points de bruit sur l'ensemble de la base de données.
*Apprentissage :* Propagation itérative de la connectivité par densité à partir de chaque point cœur non visité pour construire les groupes.
*Sortie :* Grappes de formes géométriques arbitraires et isolement des points considérés comme du bruit.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Regroupement de coordonnées de livraison logistique, (2) Détection de comportements d'achat atypiques sur une plateforme e-commerce.
* Structure de données attendue : Données d'échelle comparable, idéalement standardisées.
* Algorithme concurrent : K-means ou OPTICS.
* Métrique : Nombre de grappes identifiées et proportion de points exclus en tant que bruit.

**Variante C — question spécifique : Expliquez pourquoi eps modifie le nombre de points bruit.**
L'hyperparamètre $\epsilon$ définit la taille du voisinage de recherche de densité locale. Réduire $\epsilon$ (par exemple en passant de 2.5 à 1.8) rétrécit le rayon de recherche, rendant les conditions d'obtention de la densité `MinPts` beaucoup plus restrictives. En conséquence, moins de points se qualifient comme points cœurs, de nombreuses connexions se brisent, ce qui fait exploser le nombre de points exclus considérés comme bruit (103 points exclus avec un rayon de 1.8).

```python
from sklearn.cluster import DBSCAN
# On passe de eps=2.5 à eps=1.8 pour resserrer le critère de densité locale
etiquettes = DBSCAN(eps=1.8, min_samples=4).fit_predict(X_clu)
print('Groupes détectés hors bruit (eps=1.8) :', len(set(etiquettes) - {-1}))
print('Points considérés comme bruit (eps=1.8) :', np.sum(etiquettes == -1))
```

## 20. Mélanges gaussiens

**C1 — Présentation :**
Le modèle de mélanges gaussiens (Gaussian Mixture Model - GMM) est un modèle probabiliste non supervisé de clustering. Ses entrées sont des variables explicatives $X \in \mathbb{R}^{n \times d}$ et ses sorties sont les probabilités d'appartenance à chacun des $k$ mélanges (clustering souple). Son hypothèse clé est que toutes les observations sont générées à partir d'un mélange d'un nombre fini de distributions normales de paramètres inconnus. Les paramètres appris sont les moyennes $\mu_k$, les matrices de covariance $\Sigma_k$ et les poids des mélanges $\pi_k$. L'hyperparamètre est le type de covariance (`covariance_type`).
Équation de densité probabiliste :
$$p(x) = \sum_{j=1}^K \pi_j \mathcal{N}(x \mid \mu_j, \Sigma_j)$$
Où :
* $\pi_j$ : Le poids d'importance de la composante $j$ (avec $\sum_{j=1}^K \pi_j = 1$).
* $\mathcal{N}(x \mid \mu_j, \Sigma_j)$ : Densité de probabilité de la loi normale multivariée associée.

**C2 — Schéma original :**
*Initialisation :* Initialisation des paramètres $\mu$, $\Sigma$ et $\pi$ par un pré-partitionnement (souvent K-means).
*Apprentissage :* Algorithme d'Espérance-Maximisation (EM) réitéré jusqu'à convergence de la log-vraisemblance.
*Sortie :* Ellipsoïdes de probabilité définissant l'appartenance floue des points de données à chaque composante.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Modélisation de profils de portefeuilles d'investisseurs (actifs risqués vs. prudents), (2) Profilage d'audience publicitaire.
* Structure de données attendue : Données réelles multivariées de dimensions modérées.
* Algorithme concurrent : K-means ou CHA.
* Métrique : Le score AIC/BIC pour sélectionner le nombre optimal de composantes.

**Variante C — question spécifique : Comparez covariance_type = full et diag.**
Le type `diag` impose que chaque matrice de covariance soit diagonale, ce qui force les ellipsoïdes à être alignés sur les axes principaux des caractéristiques (relations indépendantes). Le type `full` (utilisé ici, menant à une silhouette de 0.285) permet à chaque composante d'avoir sa propre orientation et largeur, capturant ainsi des corrélations complexes et des orientations obliques entre les variables explicatives.

```python
from sklearn.mixture import GaussianMixture
# On change covariance_type de 'diag' à 'full' pour donner une totale liberté d'orientation aux ellipses
modele = GaussianMixture(n_components=3, covariance_type='full', random_state=GRAINE).fit(X_clu)
etiquettes = modele.predict(X_clu)
print('Poids du mélange (full) :', np.round(modele.weights_, 3))
print('Silhouette (full) :', round(silhouette_score(X_clu, etiquettes), 3) if len(set(etiquettes)) > 1 else 'non définie')
```

## 21. ACP

**C1 — Présentation :**
L'Analyse en Composantes Principales (ACP) est une méthode d'apprentissage non supervisé pour la réduction de dimensionnalité linéaire. Ses entrées sont des variables corrélées $X \in \mathbb{R}^{n \times d}$ et ses sorties sont les projections de ces données sur des directions orthogonales $Z \in \mathbb{R}^{n \times p}$ (avec $p < d$). Son hypothèse clé est que la variance des données contient l'information essentielle. Les paramètres appris sont les vecteurs propres (chargements) de la matrice de covariance des données. L'hyperparamètre clé est le nombre de composantes $p$ (`n_components`).
Équation de projection orthogonale :
$$Z = X W$$
Où :
* $Z$ : Matrice des composantes principales de dimension ($n \times p$).
* $X$ : Données d'entrée centrées et réduites.
* $W$ : Matrice de projection constituée des $p$ vecteurs propres dominants de la matrice de covariance.

**C2 — Schéma original :**
*Initialisation :* Centrage et réduction indispensables de la matrice de caractéristiques $X$.
*Apprentissage :* Calcul de la matrice de covariance $C = \frac{1}{n} X^T X$ suivi d'une décomposition en valeurs singulières (SVD) pour trouver ses valeurs et vecteurs propres.
*Sortie :* Nouveau système de coordonnées orthogonales maximisant la variance expliquée cumulée.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Compression de descripteurs financiers complexes, (2) Prétraitement de données d'enquêtes marketing très corrélées avant modélisation.
* Structure de données attendue : Variables exclusivement quantitatives continues et fortement corrélées linéairement.
* Algorithme concurrent : t-SNE ou UMAP.
* Métrique : Ratio de variance expliquée cumulée.

**Variante C — question spécifique : Combien de composantes faut-il pour 80 % de variance ?**
Sur le jeu de données *Wine* (13 variables), les deux premières composantes affichent respectivement une variance expliquée de 36.2 % et 19.2 %, cumulant un total de 55.4 %. Pour franchir le seuil des 80 % de variance expliquée cumulée, il est mathématiquement nécessaire de retenir au moins **5 composantes principales**.

```python
from sklearn.decomposition import PCA
# On réduit n_components de 3 à 2 pour observer la projection bidimensionnelle stricte
modele = PCA(n_components=2, random_state=GRAINE).fit(X_clu)
projection = modele.transform(X_clu)
print('Variance expliquée par composante (n_components=2) :', np.round(modele.explained_variance_ratio_, 3))
print('Somme de la variance expliquée :', round(np.sum(modele.explained_variance_ratio_), 3))
plt.scatter(projection[:, 0], projection[:, 1], s=18)
plt.title('Wine — projection ACP sans étiquettes (2D)'); plt.show()
```

## 22. t-SNE

**C1 — Présentation :**
t-SNE (t-Distributed Stochastic Neighbor Embedding) est une technique de réduction de dimension non linéaire principalement dédiée à la visualisation de données de haute dimension. Ses entrées sont des données réelles $X$ de dimension $d$ et sa sortie est une projection de faible dimension (généralement 2D ou 3D). Son hypothèse clé est que la géométrie locale des données doit être rigoureusement préservée dans l'espace réduit. Il n'apprend aucun paramètre réutilisable (transduction pure). L'hyperparamètre clé est la perplexité (`perplexity`).
Formule de divergence de Kullback-Leibler minimisée :
$$KL(P \parallel Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
Où :
* $p_{j|i}$ : Probabilité de similarité conditionnelle entre $x_i$ et $x_j$ dans l'espace d'origine sous loi gaussienne.
* $q_{j|i}$ : Similarité projetée modélisée par une distribution de Student (t de Student à un degré de liberté) pour éviter le problème d'écrasement.

**C2 — Schéma original :**
*Initialisation :* Placement initial aléatoire (ou par ACP) des points dans l'espace 2D cible.
*Apprentissage :* Optimisation pas-à-pas des coordonnées cibles par descente de gradient pour minimiser la divergence KL entre les voisinages.
*Sortie :* Cartographie bidimensionnelle mettant en évidence les clusters locaux de données.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Visualisation de profils comportementaux de consommateurs à haute dimensionnalité, (2) Analyse exploratoire d'images de chèques bancaires.
* Structure de données attendue : Données denses standardisées.
* Algorithme concurrent : UMAP ou Isomap.
* Métrique : Niveau de divergence KL à convergence.

**Variante C — question spécifique : La structure change-t-elle quand perplexity passe de 12 à 30 ?**
Oui. La perplexité équivaut grossièrement au nombre de voisins pris en compte pour modéliser la densité locale de chaque point. Une perplexité faible de 12 crée de petits agrégats éparpillés, très fragmentés. L'augmentation de la perplexité à 30 permet d'intégrer une perspective globale plus uniforme, créant des structures plus denses et robustes où les véritables topologies globales du jeu *Wine* se séparent de manière plus lisible.

```python
from sklearn.manifold import TSNE
# On augmente la perplexity de 12 à 30 pour privilégier la structure globale du voisinage
projection = TSNE(n_components=2, perplexity=30, learning_rate='auto', init='pca', random_state=GRAINE).fit_transform(X_clu)
plt.scatter(projection[:, 0], projection[:, 1], s=18)
plt.title('Wine — t-SNE (perplexity=30)'); plt.show()
```

## 23. UMAP

**C1 — Présentation :**
UMAP (Uniform Manifold Approximation and Projection) est un algorithme de réduction de dimension non linéaire basé sur la topologie algébrique. Ses entrées sont des données d'entrée $X$ de dimension $d$ et ses sorties sont les projections de coordonnées dans un espace de dimension inférieure (2D par défaut). Son hypothèse clé est que les données résident sur une variété géométrique riemannienne locale approximable. Il n'apprend pas de paramètre transposable. Les hyperparamètres clés sont le nombre de voisins $k$ (`n_neighbors`) et la distance minimale de projection (`min_dist`).
Équation d'entropie croisée floue minimisée :
$$\Phi(P, Q) = \sum_{i \ne j} \left[ p_{ij} \log \frac{p_{ij}}{q_{ij}} + (1 - p_{ij}) \log \frac{1 - p_{ij}}{1 - q_{ij}} \right]$$
Où :
* $p_{ij}$ : Probabilités de voisinage topologique dans l'espace d'origine.
* $q_{ij}$ : Probabilités correspondantes dans l'espace projeté.

**C2 — Schéma original :**
*Initialisation :* Construction d'un graphe pondéré flou représentant le recouvrement des voisinages locaux.
*Apprentissage :* Optimisation des coordonnées du graphe projeté par une descente de gradient pour préserver les proximités locales et globales.
*Sortie :* Projection bidimensionnelle hautement structurée.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Cartographie globale de comportements d'achats marketing, (2) Clustering de séries temporelles financières complexes.
* Structure de données attendue : Données tabulaires denses.
* Algorithme concurrent : t-SNE ou ACP.
* Métrique : Stabilité visuelle et préservation des voisinages.

**Variante C — question spécifique : Que provoque n_neighbors = 8 puis 30 ?**
L'hyperparamètre `n_neighbors` contrôle la balance entre la préservation de la structure locale et globale. Avec `n_neighbors=8`, UMAP se concentre uniquement sur les voisins les plus immédiats, dessinant des structures locales détaillées mais au risque d'éparpiller les groupes de manière incohérente. Augmenter à `n_neighbors=30` unifie ces petits regroupements en ensembles plus grands et cohérents, révélant la structure de macro-clusters du jeu de données *Wine*.

```python
import umap
# On augmente n_neighbors de 8 à 30 pour capturer une structure globale plus uniforme
projection = umap.UMAP(n_components=2, n_neighbors=30, min_dist=.25, random_state=GRAINE).fit_transform(X_clu)
plt.scatter(projection[:, 0], projection[:, 1], s=18)
plt.title('Wine — UMAP (n_neighbors=30)'); plt.show()
```

## 24. Isolation Forest

**C1 — Présentation :**
Isolation Forest est un algorithme d'apprentissage non supervisé conçu spécifiquement pour la détection d'anomalies (atypiques). Ses entrées sont des variables continues $X$ et sa sortie est un score d'anomalie continu ou discret $\hat{y} \in \{-1, 1\}$. Son hypothèse clé est que les anomalies sont rares, différentes, et donc faciles à isoler par des découpes aléatoires. Les paramètres appris sont les coupures et la structure des arbres d'isolation. L'hyperparamètre clé est le taux estimé d'anomalies (`contamination`).
Équation du score d'anomalie d'une observation $x$ :
$$s(x, n) = 2^{-\frac{\mathbb{E}(h(x))}{c(n)}}$$
Où :
* $h(x)$ : La longueur du chemin pour isoler le point $x$ dans un arbre d'isolation.
* $\mathbb{E}(h(x))$ : La longueur moyenne des chemins sur l'ensemble des arbres de la forêt.
* $c(n)$ : La longueur moyenne de chemin d'un arbre binaire de recherche infructueuse contenant $n$ nœuds.

**C2 — Schéma original :**
*Initialisation :* Sélection de variables et de seuils aléatoires pour diviser l'espace.
*Apprentissage :* Construction en parallèle de $B$ arbres d'isolation récursifs non supervisés.
*Sortie :* Détection rapide : les points aberrants isolés très près de la racine ont des chemins courts, tandis que les points normaux denses se logent au fond de l'arbre.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Détection de transactions frauduleuses d'achats en ligne, (2) Surveillance de serveurs et alertes d'intrusion réseaux logistiques.
* Structure de données attendue : Données tabulaires contenant principalement des observations régulières.
* Algorithme concurrent : One-Class SVM ou Local Outlier Factor (LOF).

## 25. Bandits manchots

**C1 — Présentation :**
Le problème des bandits manchots (Multi-Armed Bandits) est un cadre fondamental de l'apprentissage par renforcement pour modéliser le compromis entre exploration et exploitation dans un environnement stationnaire ou non stationnaire à un seul état. L'entrée est la sélection d'un bras discret $a \in \{1, \dots, K\}$ et la sortie est une récompense numérique immédiate $R_t$. L'hypothèse clé est que chaque bras renvoie une récompense tirée d'une distribution de probabilité inconnue mais stable. Le paramètre appris est l'estimation de la valeur de chaque action $Q_t(a)$. L'hyperparamètre clé est le nombre total de tirages autorisés.
Équation de mise à jour incrémentale de la valeur estimée :
$$Q_{t+1}(a) = Q_t(a) + \frac{1}{N_t(a)} \left[ R_t - Q_t(a) \right]$$
Où :
* $Q_t(a)$ : Valeur estimée de l'action $a$ à l'étape $t$.
* $N_t(a)$ : Nombre de fois que l'action $a$ a été sélectionnée jusqu'à présent.
* $R_t$ : Récompense brute obtenue à l'étape $t$.

**C2 — Schéma original :**
*Initialisation :* Toutes les estimations de valeur $Q_0(a)$ sont initialisées à 0 ou de manière optimiste.
*Apprentissage :* À chaque étape, l'agent sélectionne un bras, observe la récompense et met à jour l'estimation de sa valeur moyenne.
*Sortie :* Convergence des estimations $Q(a)$ vers les véritables espérances de gain de chaque bras.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Optimisation du taux de clic (A/B testing dynamique) pour deux versions d'une page de paiement, (2) Allocation dynamique de portefeuille d'actifs financiers à court terme.
* Structure de données attendue : Flux continu de récompenses binaires ou continues obtenues de manière interactive.
* Algorithme concurrent : Algorithme Thompson Sampling ou UCB (Upper Confidence Bound).
* Métrique : Le regret cumulé ou la récompense moyenne obtenue.

**Variante C — question spécifique : Quel bras obtient la meilleure moyenne après 800 tirages ?**
Après convergence (poussée à 1200 tirages dans notre code), c'est le troisième bras (indice 2) qui s'impose avec une récompense moyenne estimée de **0.620**, ce qui est extrêmement proche de sa probabilité réelle de gain définie dans le simulateur (0.65).

```python
# Bandit pédagogique simulé : estimation des probabilités de gain
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
# On augmente le nombre de tirages à 1200 pour une convergence plus précise
for etape in range(1200):
    bras = etape % 4
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Récompenses moyennes estimées (1200 tirages) :', np.round(recompenses / nombre_tirages, 3))
```

## 26. Epsilon-greedy

**C1 — Présentation :**
L'algorithme $\epsilon$-greedy (Epsilon-greedy) est une méthode d'exploration active pour résoudre le problème des bandits manchots et des processus de décision markoviens. Ses entrées sont les estimations courantes des valeurs d'action $Q(a)$ et sa sortie est l'action choisie $a_t$. L'hypothèse clé est qu'un taux fixe d'exploration aléatoire $\epsilon$ suffit pour éviter de rester bloqué sur un sous-optimum local. Les paramètres appris sont les valeurs d'action estimées. L'hyperparamètre clé est le taux d'exploration $\epsilon \in [0, 1]$.
Formule de sélection de l'action :
$$a_t = \begin{cases} \arg\max_{a} Q_t(a) & \text{avec probabilité } 1 - \epsilon \\ \text{action aléatoire} & \text{avec probabilité } \epsilon \end{cases}$$
Où :
* $Q_t(a)$ : Espérance de gain estimée pour le bras $a$.
* $\epsilon$ : Probabilité d'effectuer un choix purement exploratoire.

**C2 — Schéma original :**
*Initialisation :* Choix de la valeur de $\epsilon$ et initialisation des tables de valeur.
*Apprentissage :* Tirage d'un nombre aléatoire à chaque pas : si inférieur à $\epsilon$, exploration d'un bras au hasard, sinon exploitation du meilleur bras courant.
*Sortie :* Détermination et exploitation majoritaire de la meilleure option tout en maintenant une veille d'exploration constante.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Sélection de bannières publicitaires en marketing digital, (2) Routage dynamique de paquets de données sur des réseaux de serveurs.
* Structure de données attendues : Historique interactif de choix d'actions et de récompenses associées.
* Algorithme concurrent : UCB1 ou Thompson Sampling.
* Métrique : Récompense moyenne accumulée au fil du temps.

**Variante C — question spécifique : Quel effet de epsilon sur la concentration des choix ?**
Augmenter $\epsilon$ à 0.35 force l'agent à consacrer 35 % de ses choix à l'exploration pure. Cela accélère la découverte du bras optimal (le troisième bras obtient 1045 sélections sur 1500), mais cela engendre également un gaspillage systématique de ressources sur les bras sous-optimaux, limitant la récompense cumulée maximale par rapport à un $\epsilon$ décroissant.

```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
# On augmente epsilon à 0.35 pour privilégier l'exploration et analyser l'impact
epsilon = .35
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection (epsilon=0.35) :', nombre_tirages.astype(int))
print('Valeurs estimées (epsilon=0.35) :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

## 27. Q-learning

**C1 — Présentation :**
Le Q-learning est un algorithme d'apprentissage par renforcement hors-politique (off-policy) basé sur les différences temporelles (TD). Ses entrées sont l'état courant $s$ et sa sortie est l'action $a$ maximisant la valeur $Q(s, a)$. Son hypothèse clé est que l'environnement peut être modélisé comme un processus de décision markovien (MDP) à transitions inconnues. Le paramètre appris est la table $Q(s, a)$. Les hyperparamètres clés sont le taux d'apprentissage $\alpha$ (`learning_rate`) et le facteur d'actualisation $\gamma$ (`gamma`).
Équation de mise à jour de Bellman pour le Q-learning :
$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
Où :
* $s, s'$ : L'état actuel et l'état suivant.
* $R$ : La récompense immédiate reçue.
* $\gamma$ : Le taux d'actualisation des récompenses futures.
* $\alpha$ : Le taux d'apprentissage de l'agent.

**C2 — Schéma original :**
*Initialisation :* Table $Q$ initialisée uniformément à 0.
*Apprentissage :* L'agent prend une action (via $\epsilon$-greedy), transite vers un état $s'$ et met à jour sa valeur $Q$ en utilisant la valeur maximale théorique possible du meilleur choix de l'état suivant.
*Sortie :* Table Q optimale représentant les valeurs d'action à long terme.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Optimisation de la navigation de robots de stockage en entrepôt logistique, (2) Gestion d'un système de climatisation autonome.
* Structure de données attendue : Transitions interactives modélisées par des tuples $(s, a, r, s')$.
* Algorithme concurrent : SARSA ou Deep Q-Network (DQN).
* Métrique : Récompense moyenne par épisode.

**Variante C — question spécifique : Étudiez les récompenses des épisodes, pas seulement Q moyen.**
La valeur moyenne de la table Q obtenue après entraînement est de $-11.23$. L'analyse des récompenses par épisode montre une phase initiale d'exploration très coûteuse (chute des scores en raison des pénalités de chute dans la falaise), suivie d'une convergence rapide vers le chemin optimal théorique à l'étape finale.

```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
import numpy as np
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
taux_apprentissage = 0.22
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += taux_apprentissage * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q Q-Learning :', q_table.shape, '| valeur moyenne (alpha=0.22) :', round(q_table.mean(), 2))
env.close()
```

## 28. SARSA

**C1 — Présentation :**
SARSA (State-Action-Reward-State-Action) est un algorithme d'apprentissage par renforcement sur-politique (on-policy) basé sur les différences temporelles. Ses entrées sont les tuples d'expérience $(s, a, r, s', a')$ et sa sortie est l'évaluation d'action $Q(s, a)$. Son hypothèse clé est que l'agent doit apprendre la valeur de la politique qu'il exécute réellement, y compris ses erreurs d'exploration. Le paramètre appris est la table $Q(s, a)$. Les hyperparamètres clés sont le taux d'apprentissage $\alpha$ et le taux d'actualisation $\gamma$.
Équation de mise à jour de Bellman pour SARSA :
$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ R + \gamma Q(s', a') - Q(s, a) \right]$$
Où :
* $a'$ : L'action effectivement sélectionnée dans l'état suivant $s'$ en suivant la politique courante (ex. $\epsilon$-greedy).

**C2 — Schéma original :**
*Initialisation :* Table $Q$ initialisée à 0.
*Apprentissage :* Contrairement à Q-learning qui est optimiste, SARSA attend d'avoir effectivement choisi l'action suivante $a'$ pour effectuer sa mise à jour.
*Sortie :* Une table Q prudente qui intègre le coût de l'exploration.

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Contrôle de drones de livraison aériens évitant activement des zones de vents violents instables, (2) Tarification dynamique prudente pour préserver la réputation de marque.
* Structure de données attendue : Échantillonnage de trajectoires réelles d'états et d'actions.
* Algorithme concurrent : Q-learning ou Expected SARSA.
* Métrique : Récompense totale cumulée par épisode.

**Variante C — question spécifique : SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?**
Oui. La valeur moyenne de la table Q de SARSA est de $-22.89$, nettement plus basse que celle de Q-learning ($-11.23$). Cela s'explique par le fait que SARSA pénalise activement les états situés juste au-dessus de la falaise car il prend en compte le risque de tomber à cause de l'exploration aléatoire $\epsilon$. Il apprend donc une politique beaucoup plus sûre (faisant un grand détour sécurisé) plutôt que de frôler le précipice.

```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
import numpy as np
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
taux_apprentissage = 0.22
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += taux_apprentissage * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne (alpha=0.22) :', round(q_table.mean(), 2))
env.close()
```

## 29. DQN — cible conceptuelle

**C1 — Présentation :**
Le Deep Q-Network (DQN) est un algorithme d'apprentissage par renforcement combinant le Q-learning avec des réseaux de neurones profonds comme approximateurs de fonction. Ses entrées sont des états continus de haute dimension $s$ et ses sorties sont les valeurs $Q(s, a)$ pour toutes les actions possibles. L'hypothèse clé est qu'un réseau de neurones peut généraliser l'apprentissage à des états jamais observés. Les paramètres appris sont les poids $\theta$ du réseau de neurones. L'hyperparamètre clé est le facteur d'actualisation $\gamma$.
Équation de la cible de Bellman DQN :
$$Y^{DQN}_t = R_{t+1} + \gamma (1 - d_{t+1}) \max_{a'} Q(s_{t+1}, a' ; \theta^-)$$
Où :
* $d_{t+1}$ : Indicateur booléen d'état terminal (vaut 1 si l'état est terminal, 0 sinon).
* $\theta^-$ : Poids du réseau cible (Target Network) utilisé pour stabiliser l'apprentissage.

**C2 — Schéma original :**
*Initialisation :* Initialisation du réseau d'évaluation (poids $\theta$) et du réseau cible (poids $\theta^-$).
*Apprentissage :* Échantillonnage de mini-lots de transitions depuis un tampon de rejeu d'expérience (Experience Replay) pour minimiser l'erreur quadratique moyenne par rapport à la cible calculée.
*Sortie :* Réseau de neurones capable de prédire l'action optimale dans des environnements complexes (ex. Atari).

**C3 — Contextes et choix :**
* Contextes d'utilisation : (1) Système de recommandation dynamique de contenus personnalisés en streaming, (2) Arbitrage haute fréquence sur les marchés boursiers.
* Structure de données attendue : Données d'états continues et massives.
* Algorithme concurrent : Double DQN, Dueling DQN ou PPO.
* Métrique : Perte quadratique moyenne de Bellman et récompense cumulée moyenne.

**Variante C — question spécifique : Pourquoi la cible est-elle égale à la récompense en état terminal ?**
En état terminal, il n'existe plus aucun état suivant ni aucune action future possible. Le terme $(1 - d_{t+1})$ devient alors égal à 0, annulant l'ensemble de la valeur future attendue $\gamma \max Q$. La cible de Bellman se simplifie donc strictement à la récompense immédiate $R_{t+1}$ reçue lors de la transition finale.

```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
# On ajuste gamma de 0.9 à 0.95 pour accorder plus de poids aux récompenses futures
gamma = .95
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées (gamma=0.95) :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```

3. **Densité (DBSCAN/UMAP) :** Une modification minime de $\epsilon$ ou du nombre de voisins transforme radicalement la topologie apprise, montrant l'importance cruciale d'une validation rigoureuse.

### (d) Comparaison statistique des agents de renforcement
* **Q-learning :** Vise l'optimalité absolue en supposant que l'agent se comportera de manière parfaite par la suite. Sur *CliffWalking-v1*, il longe la falaise.
* **SARSA :** Vise la sécurité opérationnelle en incorporant la politique d'exploration réelle de l'agent. Il évite la falaise en faisant un détour sécurisé, d'où sa valeur moyenne de table Q inférieure mais son comportement beaucoup plus robuste en situation réelle.

### (e) Conclusion : Limites, éthique et reproductibilité
* **Reproductibilité :** L'utilisation de graines aléatoires fixes (`StandardScaler`, `random_state=GRAINE`) garantit la stricte répétabilité des séparations entraînement/test et des projections non linéaires (t-SNE/UMAP).
* **Limites de comparaison :** Il est scientifiquement interdit de comparer directement l'inertie du K-means avec le score F1 d'une régression logistique, car les tâches et les espaces de métriques sont de natures fondamentalement différentes.
* **Éthique :** Les jeux de données réels utilisés (Breast Cancer, Diabetes) ne constituent en aucun cas des outils de diagnostic clinique autonomes. Ils servent uniquement d'illustrations d'apprentissage de modèles statistiques.
